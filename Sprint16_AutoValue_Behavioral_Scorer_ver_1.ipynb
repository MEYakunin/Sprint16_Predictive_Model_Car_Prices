Компания переходит от ручной работы штата из 50 экспертов-оценщиков к полной автоматизации процесса мгновенного выкупа. Текущая бизнес-модель требует расчета цены за одну секунду на основе данных анкеты, с последующей физической проверкой состояния авто техником в течение 15 минут. Существующий подход не масштабируется до целевых показателей (обработка 10 000 заявок в день), страдает от человеческого фактора и высокой скорости оттока клиентов.

````python
# Функция вывода графика гистограммы
def print_histplot(df, column, df_name):
    """
    Строит гистограмму распределения значений столбца с наложенной KDE-кривой,
    а также добавляет вертикальные линии для среднего и медианы с подписями.
    """
    fig, ax = plt.subplots(figsize=(12, 6))

    # Гистограмма с KDE 
    sns.histplot(df[column], bins=50, kde=True, 
                color='steelblue', edgecolor='black', label='Гистограмма')

    # Получаем линию KDE и меняем её цвет
    for line in ax.lines:
        line.set_color('red')
        line.set_linewidth(2)
        line.set_label('KDE')

    # Добавляем вертикальные линии для среднего и медианы
    ax.axvline(df[column].mean(), color='green', linestyle='--', linewidth=2, 
            label=f'Среднее: {df[column].mean():,.0f} руб.')
    ax.axvline(df[column].median(), color='orange', linestyle='--', linewidth=2, 
            label=f'Медиана: {df[column].median():,.0f} руб.')

    # Подписываем оси и добавляем заголовок
    ax.set_xlabel('Цена (руб.)', fontsize=15)
    ax.set_ylabel('Плотность', fontsize=15)
    ax.set_title(f'Распределение цен автомобилей для данных {df_name}', fontsize=20, fontweight='bold')
    ax.legend(fontsize=11)

    plt.tight_layout()
    plt.show()
````

````python
# Функция вывода графика ящик с усами
def print_boxplot(df, column, df_name):
    """
    Строит boxplot («ящик с усами») для заданного столбца DataFrame и добавляет
    аннотации с ключевыми статистиками: медианой, Q1 и Q3.
    """

    fig, ax = plt.subplots(figsize=(10, 6))

    # Boxplot (ящик с усами)
    boxplot = ax.boxplot(df[column], vert=True, patch_artist=True,
                        boxprops=dict(facecolor='lightblue', color='navy'),
                        medianprops=dict(color='red', linewidth=2),
                        whiskerprops=dict(color='navy', linewidth=2),
                        capprops=dict(color='navy', linewidth=2),
                        flierprops=dict(markerfacecolor='red', marker='o', markersize=6))

    ax.set_ylabel('Цена (руб.)', fontsize=15)
    ax.set_title(f'Boxplot распределения цен автомобилей для данных {df_name}', fontsize=20, fontweight='bold')
    ax.grid(True, alpha=0.3, axis='y')

    # Вычисляем ключевые статистики для аннотаций
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    median = df[column].median()

    # Размещаем текстовые аннотации справа от ящика
    ax.text(1.15, median, f'Медиана: {median:,.0f}', va='center', fontsize=11)
    ax.text(1.15, Q1, f'Q1: {Q1:,.0f}', va='center', fontsize=11)
    ax.text(1.15, Q3, f'Q3: {Q3:,.0f}', va='center', fontsize=11)

    plt.tight_layout()
    plt.show()
````

### [4.1 Сравнение базовых моделей градиентного бустинга <a id=4_1></a>](#content)

````python
# Создаем пустой датафрейм с метриками
metrics_result_df = pd.DataFrame(columns=['RMSE', 'MAE', 'R2', 'Overpricing_rate', 'Underpricing_loss', 'wall_time'])

# Создаем пустой словарь для хранения обученых моделей
all_model_dict = {}

# Функция для добавления метрик в список
def add_metrics(metrics_result_df, model_name, y_true, y_pred, duration):

    # Рассчитываем завышение цен на 20%
    error_ratio = (y_pred - y_true) / y_true
    overpricing_rate = (error_ratio > 0.20).mean()

    # Рассчитываем занижения цен на 20%
    under_mask = error_ratio < -0.2
    underpricing_loss = (y_true[under_mask] - y_pred[under_mask]).sum()

    # Расчет всех метрик
    metrics = {
        'RMSE': round(root_mean_squared_error(y_true, y_pred), 2),
        'MAE': round(mean_absolute_error(y_true, y_pred), 2),
        'R2': round(r2_score(y_true, y_pred), 2),
        'Overpricing_rate': round(overpricing_rate, 2),
        'Underpricing_loss': round(underpricing_loss, 2),
        'wall_time': round(duration, 2)
    }
    
    # Добавляем новую строку в results_df
    metrics_result_df.loc[model_name] = metrics
    metrics_result_df.sort_values(by='RMSE', ascending=True, inplace=True)
````

````python
def fit_base_model(model, name_model):
    """
    Обучает базовую ML-модель, фиксирует время обучения, рассчитывает прогноз
    на валидационной выборке и сохраняет метрики и саму модель.
    """
    start = time.time()
    
    model.fit(X_train, y_train, eval_set=[(X_val, y_val)])
    y_pred = model.predict(X_val)

    # Вычисляем время модели
    duration = time.time() - start

    # Выводим значения метрик
    add_metrics(metrics_result_df, name_model, y_val, y_pred, duration)
    all_model_dict[name_model] = model
````

````python
# Обучаем базовую модель XGBRegressor
model_xgboost_base = XGBRegressor(random_state=RANDOM_STATE, enable_categorical=True, verbose=1)
model_xgboost = fit_base_model(model_xgboost_base, 'XGBoost_Base')

# Обучаем базовую модель LGBMRegressor
model_lgbm_base = LGBMRegressor(random_state=RANDOM_STATE, verbose=1)
model_lgbm_base = fit_base_model(model_lgbm_base, 'LGBMRegressor_Base')

# Обучаем базовую модель CatBoost
model_catboost_base = CatBoostRegressor(random_state=RANDOM_STATE, cat_features=categorycal_cols, verbose=1)
model_catboost_base = fit_base_model(model_catboost_base, 'CatBoost_Base')

# Выводим метрики базовых моделей
display(metrics_result_df)
````

### [4.2 Анализ использования Quantile Loss (квантильной регрессии) в CatBoost <a id=2_4></a>](#content)

````python
result = []

# Подбор оптимального квантиля: перебираем alpha от 0.1 до 0.9 с шагом 0.1
for alpha in np.arange(0.1, 1.0, 0.1):
    alpha = round(alpha, 1)

    # Инициализация модели CatBoost с квантильной функцией потерь
    model_catboost_quan_loss = CatBoostRegressor(
        random_state=RANDOM_STATE,
        loss_function=f'Quantile:alpha={alpha}',
        cat_features=categorycal_cols,
        verbose=0
    )

    # Обучение и прогноз
    model_catboost_quan_loss.fit(X_train, y_train)
    y_pred_cb_ql = model_catboost_quan_loss.predict(X_val)

    # Расчёт относительного отклонения и бизнес‑метрик
    error_ratio = (y_pred_cb_ql - y_val) / y_val
    overpricing_rate = (error_ratio > 0.20).mean()

    under_mask = error_ratio < -0.2
    underpricing_loss = (y_val[under_mask] - y_pred_cb_ql[under_mask]).sum()

    # Сбор ключевых метрик для текущего alpha
    metrics = {
        'alpha': alpha,
        'RMSE': round(root_mean_squared_error(y_val, y_pred_cb_ql), 2),
        'MAE': round(mean_absolute_error(y_val, y_pred_cb_ql), 2),
        'R2': round(r2_score(y_val, y_pred_cb_ql), 2),
        'Overpricing_rate': round(overpricing_rate, 2),
        'Underpricing_loss': round(underpricing_loss, 2)
    }

    result.append(metrics)

# Формирование и фильтрация результатов: оставляем модели с долей переоценки ≤ 20%, сортируем по RMSE
result_df = pd.DataFrame(result)
result_df_filt = (
    result_df[result_df['Overpricing_rate']<=0.2]
    .sort_values(by=['RMSE'], ascending=True)
)

best_alpha = result_df_filt['alpha'].iloc[0]
display(result_df_filt)
````

````python
# Обучаем базовую модель CatBoost c Quantile Loss
model_catboost_base_quan = CatBoostRegressor(
    random_state=RANDOM_STATE, 
    loss_function=f'Quantile:alpha={best_alpha}',
    cat_features=categorycal_cols, 
    verbose=0
)

model_catboost_base_quan = fit_base_model(model_catboost_base_quan, f'CatBoost_Base_Quantile:alpha={best_alpha}')

# Выводим метрики базовых моделей
display(metrics_result_df)
````

## [Этап 5. Подбор гиперпараметров (Optuna) <a id=5></a>](#content)

### [5.1 Оптимизация базовых моделей <a id=5_1></a>](#content)

````python
# Создаем словарь гиперпораметров моделей
MODEL_CONFIG = {
    'XGBoostRegressor': {
        'class': XGBRegressor,
        'params': lambda t: {
            'learning_rate': t.suggest_float('learning_rate', 0.01, 0.1, log=True),
            'max_depth': t.suggest_int('max_depth', 3, 10),
            'n_estimators': t.suggest_int('n_estimators', 500, 2000),
            'gamma': t.suggest_float('gamma', 1e-8, 1.0, log=True),
            'reg_lambda': t.suggest_float('reg_lambda', 1e-8, 10.0, log=True),
            'reg_alpha': t.suggest_float('reg_alpha', 1e-8, 10.0, log=True),
            'early_stopping_rounds': EARLY_STOPPING_ROUNDS
        },
        'fit_kwargs': { 
            'n_jobs': -1, 
            'enable_categorical': True
        }
    },
    'CatBoostRegressor': {
        'class': CatBoostRegressor,
        'params': lambda t: {
            'learning_rate': t.suggest_float('learning_rate', 0.01, 0.1, log=True),
            'depth': t.suggest_int('depth', 3, 10),
            'iterations': t.suggest_int('iterations', 500, 2000),
            'l2_leaf_reg': t.suggest_float('l2_leaf_reg', 1, 10)
        },
        'fit_kwargs': {
            'early_stopping_rounds': EARLY_STOPPING_ROUNDS,
            'cat_features': categorycal_cols}
    },
    'LGBMRegressor': {
        'class': LGBMRegressor,
        'params': lambda t: {
            'learning_rate': t.suggest_float('learning_rate', 0.01, 0.1, log=True),
            'max_depth': t.suggest_int('max_depth', 3, 10),
            'n_estimators': t.suggest_int('n_estimators', 500, 2000),
            'num_leaves': t.suggest_int('num_leaves', 20, 256),
            'min_child_samples': t.suggest_int('min_child_samples', 5, 100),
            'reg_alpha': t.suggest_float('reg_alpha', 1e-8, 1.0, log=True),
            'reg_lambda': t.suggest_float('reg_lambda', 1e-8, 1.0, log=True)
        },
        'fit_kwargs': {
            'callbacks': [lgb.early_stopping(stopping_rounds=20)], 
            'n_jobs': -1
        }
    }
}
````

````python
def objective(trial, model_name, X, y):
    """
    Целевая функция для Optuna: возвращает среднюю RMSE по кросс‑валидации.
    """

    cfg = MODEL_CONFIG[model_name]
    params = cfg['params'](trial)

    kf = KFold(n_splits=3, shuffle=True, random_state=RANDOM_STATE)
    scores = []

    for _, (train_idx, val_idx) in enumerate(kf.split(X)):
        X_tr, X_va = X.iloc[train_idx], X.iloc[val_idx]
        y_tr, y_va = y.iloc[train_idx], y.iloc[val_idx]

        model = cfg['class'](**params, **cfg['fit_kwargs'])

        model.fit(X_tr, y_tr, eval_set=[(X_va, y_va)])
            
        y_pred = model.predict(X_va)
        scores.append(root_mean_squared_error(y_va, y_pred))

    return np.mean(scores)
````

````python
# Запускем оптимизацию моделей
for name in MODEL_CONFIG:
    study = optuna.create_study(
        direction='minimize',
        sampler=optuna.samplers.TPESampler(seed=RANDOM_STATE),
        study_name=f'{name}_tuning_{int(time.time())}',
        pruner=MedianPruner(n_startup_trials=5, n_warmup_steps=10)
    )
    study.optimize(
        lambda trial, n=name: objective(trial, n, X_train, y_train),
        n_trials=15,
        n_jobs=-1,
        show_progress_bar=False
    )

    MODEL_CONFIG[name]['best_params'] = study.best_params
    MODEL_CONFIG[name]['best_score'] = study.best_value
````

````python
# Обучение оптимизированных моделей
for name in MODEL_CONFIG:
    model = MODEL_CONFIG[name]['class'](
        **MODEL_CONFIG[name]['best_params'],
        **MODEL_CONFIG[name]['fit_kwargs']
    )
    fit_base_model(model, f'{name}_Tuning')

display(metrics_result_df)
````

### [6.2 Анализ склонности к завышению и занижению стоимости <a id=6_2></a>](#content)

````python
def analyze_pricing_losses_by_group(model, X, y, name_model, groupby):
    """
    Анализирует потери из‑за переоценки и недооценки стоимости по заданной группе (например, бренд или регион).
    Вычисляет доли и абсолютные значения потерь, строит график упущенной выгоды и выводит сводную таблицу.
    """
    df_error = pd.DataFrame()
    y_pred = model.predict(X)

    # Рассчитываем относительное отклонение прогноза от факта (в долях)
    error_ratio = (y_pred - y) / y

    # Формируем таблицу ошибок с флагами и исходными значениями
    df_error[groupby] = X[groupby]
    df_error['error_ratio'] = error_ratio
    df_error['overpricing'] = error_ratio > 0.20
    df_error['underpricing'] = error_ratio < -0.20
    df_error['price_rub'] = y
    df_error['price_rub_prediction'] = y_pred
    df_error['abs_error'] = np.abs(y_pred - y)

    # Рассчитаем overpricing_loss (потери при переоценке)
    df_error['overpricing_loss'] = df_error.apply(
        lambda row: (row['price_rub_prediction'] - row['price_rub']) 
                    if row['overpricing'] 
                    else 0,
        axis=1
    )

    # Рассчитаем underpricing_loss (потери при недооценке)
    df_error['underpricing_loss'] = df_error.apply(
        lambda row: (row['price_rub'] - row['price_rub_prediction']) 
                    if row['underpricing'] 
                    else 0,
        axis=1
    )
    
    # Агрегируем метрики по группе
    df_analysis = df_error.groupby(groupby).agg(
        overpricing_rate=('overpricing', 'mean'),
        overpricing_count=('overpricing', 'sum'),
        underpricing_rate=('underpricing', 'mean'),
        underpricing_count=('underpricing', 'sum'),
        total_count=('price_rub', 'count'),
        overpricing_loss=('overpricing_loss', 'sum'),
        underpricing_loss=('underpricing_loss', 'sum'),
        mae=('abs_error', 'mean')
    ).reset_index()

    # Сортируем по убыванию упущенной выгоды (недооценка)
    df_analysis.sort_values(by='underpricing_loss', ascending=False, inplace=True)

    # Визуализация: упущенная выгода по группам
    plt.figure(figsize=(20, 10))
    sns.barplot(
        data=df_analysis,
        x='underpricing_loss',
        y=groupby,
        order=df_analysis[groupby].tolist()
    )
    plt.title(f'Упущенная выгода по {groupby} для модели {name_model}')
    plt.xlabel('Упущенная выгода (руб.)')
    plt.show()
    
    display(f'Упущенная выгода по {groupby} для модели {name_model}',df_analysis)
    print('='*150)
````

````python
# Анализ топ‑3 моделей по брендам
for name_model in metrics_result_df.head(3).index:
    model = all_model_dict[name_model]
    analyze_pricing_losses_by_group(model, X_val, y_val, name_model, 'brand')
````
